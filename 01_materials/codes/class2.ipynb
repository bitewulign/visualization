```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
fig,(ax1, ax2) = plt.subplots(ncols=2,
 nrows=1,
 figsize=(7, 3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))

# Customize scatter plot
ax1.scatter(x1, y1, color='blue', marker='o')
ax1.set_title('Scatter Plot', fontsize=14, fontweight='bold')
ax1.set_xlabel('X1', fontsize=12)
ax1.set_ylabel('Y1', fontsize=12)
ax1.annotate('Important Point', xy=(10, 60), xytext=(20, 70),
             arrowprops=dict(facecolor='black', shrink=0.05))

# Customize bar plot
ax2.bar(x2, y2, color='green')
ax2.set_title('Bar Plot', fontsize=14, fontweight='bold')
ax2.set_xlabel('X2', fontsize=12)
ax2.set_ylabel('Y2', fontsize=12)
for i, v in enumerate(y2):
    ax2.text(i, v + 5, str(v), ha='center', fontsize=10, color='black')

plt.tight_layout()
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))

# Customize scatter plot
ax1.scatter(x1, y1, color='blue', marker='o')
ax1.set_title('Scatter Plot', fontsize=14, fontweight='bold')
ax1.set_xlabel('X1', fontsize=12)
ax1.set_ylabel('Y1', fontsize=12)
ax1.annotate('Important Point', xy=(10, 60), xytext=(20, 70),
             arrowprops=dict(facecolor='black', shrink=0.05))

# Customize bar plot
ax2.bar(x2, y2, color='green')
ax2.set_title('Bar Plot', fontsize=14, fontweight='bold')
ax2.set_xlabel('X2', fontsize=12)
ax2.set_ylabel('Y2', fontsize=12)
for i, v in enumerate(y2):
    ax2.text(i, v + 5, str(v), ha='center', fontsize=10, color='black')

plt.tight_layout()

# Add borders to scatter plot
for spine in ax1.spines.values():
    spine.set_edgecolor('black')
    spine.set_linewidth(2)

# Add borders to bar plot
for spine in ax2.spines.values():
    spine.set_edgecolor('black')
    spine.set_linewidth(2)

plt.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize=(7, 4))
for label, ax in someaxes.items():
    ax.text(0.5, 0.5, f'This is {label!r}',
            fontsize=14,
            ha = 'center',
            transform=ax.transAxes)
```

```python
# first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])
# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))
# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "indigo")
ax.plot(x, y2,
color = "red")
```

```python
y2_sd = np.std(y2)


fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax.errorbar(x, y2, yerr = y2_sd, fmt = "none", ecolor= "indigo", elinewidth= 4, capsize = 6,capthick= 4 )
```

```python
from PIL import Image # to open images
import requests # to get images from URLs
from io import BytesIO # to store images
response = requests.get('https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png')
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax_image = fig.add_axes([0.1, # x coordinate (ON FIGURE, NOT AXES)
                         0.11, # y coordinate (ON FIGURE, NOT AXES)
                         0.15, # image width
                         0.35]) # image height)
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax_image = fig.add_axes([0.1, 0.11, 0.15, 0.35])
ax_image.imshow(image)
ax_image.axis('off')
plt.show()
```
